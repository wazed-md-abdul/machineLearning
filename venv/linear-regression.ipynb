```python
import numpy as np

# ১. আমাদের ডেটা (আসল সম্পর্ক: y = 2x + 1)
X = np.array([1, 2, 3, 4, 5])
y = np.array([3, 5, 7, 9, 11])

# ২. শুরুতে আমরা আন্দাজে Weight এবং Bias ধরে নিলাম
w = 0.0
b = 0.0
learning_rate = 0.01

# ৩. ১০০০ বার চেষ্টা করে Loss কমানো (Gradient Descent)
for step in range(1000):
    # বর্তমান w এবং b দিয়ে অনুমান (Prediction)
    y_pred = w * X + b
    
    # L2 Loss বা MSE হিসাব করা
    loss = np.mean((y - y_pred) ** 2)
    
    # ভুল কমানোর জন্য w এবং b কতটুকু বদলাতে হবে (Gradients)
    dw = -2 * np.mean(X * (y - y_pred))
    db = -2 * np.mean(y - y_pred)
    
    # নতুন w এবং b আপডেট করা
    w = w - learning_rate * dw
    b = b - learning_rate * db
    
    # প্রতি ২০০ ধাপ পরপর অবস্থা দেখা
    if step % 200 == 0:
        print(f"Step {step}: Loss = {loss:.4f} | w = {w:.2f}, b = {b:.2f}")

print(f"\nট্রেইনিং শেষে ফলাফল: w = {w:.2f}, b = {b:.2f}")
```
